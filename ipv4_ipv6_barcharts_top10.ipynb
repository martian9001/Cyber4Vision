```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.ticker as ticker
import humanize

# Formatter using humanize
def humanize_formatter(x, pos):
    return humanize.intword(x, format="%.1f")

# Load dataset
dataset = pd.read_csv("ip_alloc(in).csv", index_col="country name")

# Group by country and sum IPv4 allocations
ip_pr_country = dataset.groupby("country name")["ipv4"].sum().reset_index()

# Exclude 'World' row if present
ip_pr_country = ip_pr_country[ip_pr_country["country name"] != "World"]

# Get top 10 countries by IPv4 allocation
desc_countries_top_10 = ip_pr_country.sort_values(by="ipv4", ascending=False).head(10)

# Set dark background style
plt.style.use("dark_background")

# Plot
plt.figure(figsize=(12, 6))
ax = sns.barplot(x="country name", y="ipv4", data=desc_countries_top_10, palette="Blues_r")

# Labels and title
ax.set_xlabel("Country", color='white')
ax.set_ylabel("IPv4 Allocation", color='white')
ax.set_title("Top 10 Countries by IPv4 Allocation", color='white')

# Tick label colors and rotation
ax.tick_params(colors='white')
ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')

# Apply humanized y-axis labels
ax.yaxis.set_major_formatter(ticker.FuncFormatter(humanize_formatter))

plt.tight_layout()
plt.show()
```
